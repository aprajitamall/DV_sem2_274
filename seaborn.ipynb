```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame({
    "student":["101","102","103","104","105","106","107","108","109"],
    "branch":["CSE","ECE","ME","DS","CSE","CSE","ECE","CSE","ME"],
    "hours":[5,6,4,9,7,3,2,5,6],
    "marks":[56,78,99,67,87,90,70,76,72]
  
})
display(df)
```

```python
plt.figure(figsize=(7,4))
sns.countplot( data=df,x="branch",order=["CSE","ECE","ME","DS"], color="cornflowerblue")

plt.title("No of student in each branch")
plt.xlabel("Branch")
plt.ylabel("No. of students")
plt.tight_layout()
plt.show()
```

```python
df["Study_Group"]=df["hours"].apply(lambda hours:"3 hours or less" if hours <=3 else "More than 3 hours")
display(df[["student","branch","hours","Study_Group"]])
plt.figure(figsize=(9,4))
sns.countplot( data=df,x="branch",order=["CSE","ECE","ME","DS"], color="cornflowerblue")
plt.title("Student count by Branch and Study group")
plt.xlabel("Branch")
plt.ylabel("No. of students")
plt.legend(title="Study group")
plt.tight_layout()
plt.show()
```

```python
df["Study_Group"]=df["hours"].apply(lambda hours:"3 hours or less" if hours <=3 else "More than 3 hours")
display(df[["student","branch","hours","Study_Group"]])
plt.figure(figsize=(9,4))
sns.countplot( data=df,x="branch",hue="Study_Group",order=["CSE","ECE","ME","DS"], color="cornflowerblue")
plt.title("Student count by Branch and Study group")
plt.xlabel("Branch")
plt.ylabel("No. of students")
plt.legend(title="Study group")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4))

sns.barplot(
    data=df,
    x="branch",
    y="marks",
    order=["CSE", "ECE", "ME", "DS"],
    errorbar=None
)

plt.title("Student Marks by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 4))

sns.barplot(
    data=df,
    x="branch",
    y="marks",
    order=["CSE", "ECE", "ME", "DS"],
    estimator=max,
    errorbar=None
)

plt.title("Maximum Marks by Branch")
plt.xlabel("Branch")
plt.ylabel("maximum Marks")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```
