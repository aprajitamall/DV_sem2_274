```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.DataFrame({
    "student":["101","102","103","104","105","106","107","108","109","110"],
    "branch":["CSE","ECE","ME","DS","CSE","CSE","ECE","CSE","ME","DS"],
    "hours":[5,6,4,9,7,3,2,5,6,3],
    "marks":[56,78,99,67,87,90,70,76,72,89]
  
})
display(df)
```

```python
plt.figure(figsize=(7,4))
sns.countplot( data=df,x="branch",order=["CSE","ECE","ME","DS"], color="cornflowerblue")

plt.title("No of student in each branch")
plt.xlabel("Branch")
plt.ylabel("No. of students")
plt.tight_layout()
plt.show()
```

```python
df["Study_Group"]=df["hours"].apply(lambda hours:"3 hours or less" if hours <=3 else "More than 3 hours")
display(df[["student","branch","hours","Study_Group"]])
plt.figure(figsize=(9,4))
sns.countplot( data=df,x="branch",order=["CSE","ECE","ME","DS"], color="cornflowerblue")
plt.title("Student count by Branch and Study group")
plt.xlabel("Branch")
plt.ylabel("No. of students")
plt.legend(title="Study group")
plt.tight_layout()
plt.show()
```

```python
df["Study_Group"]=df["hours"].apply(lambda hours:"3 hours or less" if hours <=3 else "More than 3 hours")
display(df[["student","branch","hours","Study_Group"]])
plt.figure(figsize=(9,4))
sns.countplot( data=df,x="branch",hue="Study_Group",order=["CSE","ECE","ME","DS"], color="cornflowerblue")
plt.title("Student count by Branch and Study group")
plt.xlabel("Branch")
plt.ylabel("No. of students")
plt.legend(title="Study group")
plt.tight_layout()
plt.show()
```

```python
#Barplot
plt.figure(figsize=(9, 4))

sns.barplot(
    data=df,
    x="branch",
    y="marks",
    order=["CSE", "ECE", "ME", "DS"],
    errorbar=None
)

plt.title("Student Marks by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Marks")
plt.ylim(0, 100)

plt.tight_layout()
plt.show()
```

```python
# Voilin plot
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    x='day',
    y='total_bill',
    hue='day',
    palette='Set2',
    legend=False
)

plt.title('Distribution of Total Bills Across Different Days')
plt.xlabel('Day')
plt.ylabel('Total Bill ($)')

plt.show()
```

```python
# Scatter Plot
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('tips')

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='smoker',
    style='smoker',
    s=80
)

plt.title('Relationship Between Total Bill and Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')

plt.show()
```
