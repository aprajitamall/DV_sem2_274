```python
import plotly
import plotly.express as px
import plotly. graph_objects as go
import pandas as pd
import numpy as np
print("Plotly version:",plotly.__version__)
```

```python
data = {
    "Subject": ["Python", "Statistics", "Visualization"],
    "Marks": [85, 90, 95]
}

fig = px.bar(
    data,
    x="Subject",
    y="Marks",
    title="My First Plotly Chart"
)

fig.show()
```

```python
sales_df=pd.DataFrame({
    "Product":["Laptop","Phone","Tablet","Monitor","Keyboard"],
    "Units_Sold":[18,35,22,14,42],
    "Revenue":[18000,14000,6600,5600,2100],
    "Category":["Computer","Moblie","Moblie","Computer","Accessory"]

})
display(sales_df)
print("Shape(rows,columns):",sales_df.shape)
print("Columns:",list(sales_df.columns))
```

```python
fig=px.bar(
    sales_df,
    x="Product",
    y="Revenue",
    title="Revenue by Product",
    color="Category",
    hover_data="Units_Sold"
)
fig.show()
```

```python
fig_go=go.Figure()
fig_go.add_trace(go.Bar(
    x=sales_df["Product"],
    y=sales_df["Revenue"],
    name="Revenue",
    marker_color="teal",
    hovertemplate="plotly_white")
)
fig_go.show()
```
