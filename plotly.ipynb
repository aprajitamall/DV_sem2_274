```python
import plotly
import plotly.express as px
import plotly. graph_objects as go
import pandas as pd
import numpy as np
print("Plotly version:",plotly.__version__)
```

```python
data = {
    "Subject": ["Python", "Statistics", "Visualization"],
    "Marks": [85, 90, 95]
}

fig = px.bar(
    data,
    x="Subject",
    y="Marks",
    title="My First Plotly Chart"
)

fig.show(renderer="browser")
```
