```python
birth_year=int(input("Enter your Birth year: "))

if birth_year>=1981 and birth_year<=1996:
    print("You are a millennial.")  
elif birth_year>=1965 and birth_year<=1980:
    print("You are a Gen X.")
elif birth_year>=2000 and birth_year<=2013:
    print("You are a Gen Z.")
elif birth_year>=2014 and birth_year<=2025:
    print("You are a Gen Alpha.")
    
else:
    print("You are a Beta.")
```

```python
year=int(input("Enter the year."))
if year %4==0:
    if year%100==0:
        if year%400==0:
            print(year,"is a leap year")
        else:
            print(year,"is not a leap year")
    else:
        print(year,"is a leap year")
else:
    print(year,"is not a leap year")
```

```python
num1=float(input("Enter the no."))
num2=float(input("Enter the no."))
op=input("Enter operation(+,-,/,*):")
if op=='+':
    result=num1+num2
elif op=='-':
    result= num1-num2
elif op=='*':
    result =num1*num2 
elif op=='/':
    if num2!=0:
        result=num1/num2  
    else:
        result="Error! Devision by zero."
else:
    result="Invalid operation."  
print("Result:",result)  
```
