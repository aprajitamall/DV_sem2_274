if two or more variable having same value,then they will occupy same memory address in python.although it saves memory but its disadvantage is that in change in any on will relfect in other. so, we do it using shallow copy or deep copy.

```python
# bitwise operator
x=int(input("Enter first no:"))
y=int(input("Enter second no:"))
print("bitwise AND:",x&y)
print("bitwise OR:",x|y)
print("bitwise XOR:",x^y)
print("bitwise NOT:",~y)
print("bitwise left shift:",x<<1)
print("bitwise right shift:",x>>1)
```
