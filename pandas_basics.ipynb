### .at and .iat
 * difference between .loc and .at is that .loc can be used to access multiple elements of a Series or DataFrame, while .at can only be used to access a single element. Similarly, the difference between .iloc and .iat is that .iloc can be used to access multiple elements of a Series or DataFrame using integer-based indexing, while .iat can only be used to access a single element.


```python
print(s.iat[1],type(s.iat[1]))
```

```python
print(s.at[2002],type(s.at[2002]))
```

```python
s.iloc[-3:] #in indexing, step size determines the increment between indices. A step size of 1 means that every index is included,
# while a step size of 2 means that every other index is included, and so on. 
# A negative step size means that the indices are accessed in reverse order.
```

```python
s.iloc[::-2] ## it print in reverse direction .why? step is negative
```

```python
s.iloc[::2]
```
