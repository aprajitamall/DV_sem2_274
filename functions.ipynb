Arbitary argumaent: when we do not know the no. of arguments then we wil get to know about it after completion or at run  time.

Types :
1. simple arbitary(*args)
2. keyword arbitary (**kwargs)

```python
## simple  arbitary argument
def add(*args):
    print(type(args))
    print(args)  # the sequence stored in args is tuple.
    sum=0
    for i in args:
        sum=sum+i
    return sum   
x=add(10,12,46,9,0,11)
print("Sum is :",x)     
```

Function Vs Method.

Function is called independently. while method is called through objects and it is related to class.

```python
## Keyword arbitary argument
def add(**kwargs):
    print(type(kwargs))
    print(kwargs)  # the sequence stored in kargs is dictionary.
    sum=0
    for i in kwargs:
        sum=sum+i
    return sum   
x=add(a=10,b=12,c=46,d=9,e=0,f=11)
print("Sum is :",x)     
```

```python
# Write a program using function to generate perfect no. betweem 1 to 1000.
# example 6
def is_perfect(num):
    sum_divisors = 0

    for i in range(1, num):
        if num % i == 0:
            sum_divisors += i

    return sum_divisors == num


print("Perfect numbers between 1 and 1000 are:")

for n in range(1, 1001):
    if is_perfect(n):
        print(n)
```

```python
#Temperature conversion
def convert_temperature(temp,unit):
    if unit=='C':
        return temp*9/5 +32
    elif unit=="F":
        return (temp-32)*5/9
    else:
        return None
print(convert_temperature(25,'C'))
print(convert_temperature(77,'F'))    
```

```python
#write the python program to check the strength of password
def is_strong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
print(is_strong_password("WeakPwd"))
print(is_strong_password("Str0ng@123"))
    
```

```python
# Calculate the total cost of items in a Shopping Cart
def calculate_total_cost(cart):
    total_cost=0
    for item in cart:
        total_cost+=item['price']*item['quantity']
    return total_cost
cart=[{'name':'Apple','price':6.0,'quantity':4},
      {'name':'Banana','price':1.2,'quantity':3},
      {'name':'Mango','price':64.8,'quantity':9}
      ]    
total_cost=calculate_total_cost(cart)
print(total_cost)
```

```python
#check if a string is palindrome
def is_palindrome(s):
    print(s.capitalize())  # makes first letter of first word of sentence capital
    print(s.title())  #makes first letter of every word of sentence capital
    print(s.upper())
    print(s.lower())
    print(s.replace("a","$"))
    s=s.lower().replace(" ","")
    print(s)
    print(s[::-1])
    return s==s[::-1]
print(is_palindrome("a man a plan a canal Panana"))
print(is_palindrome("naman"))
```

```python
#Factorial using recursion
def factorial(n):
    if n==0 or n==1:
        return n
    else:
        return n*factorial(n-1)
n=int (input()) 
print(factorial(n))   
```

##### Zip() function
* In python ,zip function is used to combine two or more  iterables(like lists, tuple,strings etc)into single iterator of tuples,where each tuple contains elements from the iterables at the same position.

```python
names=["Alice","Bob","Charlie"]
scores=[87,97,45]
combine=zip(names,scores)
print(list(combine))
```

```python
a=[1,2,3]
b=['x','y']
print(list(zip(a,b)))  # extra element 3 is ignored
```

```python
# if you want until the longest iterable ,we used this package
from itertools import zip_longest
a=[1,2,3,5]
b=['x','y']
print(list(zip_longest(a,b,fillvalue="-"))) 
```

```python
# unzipping 
pairs=[('Alice',85),('bob',90)]
names,scores=zip(*pairs)
print(names)
print(scores)
```
