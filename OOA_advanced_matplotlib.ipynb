### Advanced Matplotlib

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr"]
sales=[100,150,120,200]
profit=[20,35,25,50]
fig,ax=plt.subplots(1,2, figsize=(10,4))
ax[0].plot(months,sales,marker="o")
ax[0].set_title("Monthlyu sales")
ax[0].set_xlabel("Monts")
ax[0].set_ylabel("Sales")

ax[1].bar(months,profit)
ax[1].set_title("Monthlyu profit")
ax[1].set_xlabel("Months")
ax[1].set_ylabel("Profit")

plt.show()
```

#### Twin axes
* used to display two different datasets in same plotting  area even when the dataset  use different units or scales.
* ax.twinx(),ax.twiny()

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr","May"]
sales=[100,150,120,180,220]
temperature=[20,25,23,28,32]
fig,ax1=plt.subplots(figsize=(9,5))
line1=ax1.plot(months, sales,color="blue",marker="o",label="Sales")
ax1.set_xlabel("Months")
ax1.set_ylabel("Sales",color="blue")
ax1.tick_params(axis="y",labelcolor="blue")
ax2=ax1.twinx()
line2=ax2.plot(months, temperature,color="red",marker="s",label="Temperature")

ax2.set_ylabel("Temperature(C)",color="red")
ax2.tick_params(axis="y",labelcolor="red")

lines=line1+line2
labels=(line.get_label() for line in lines)
ax1.legend(lines,labels,loc="upper left")
plt.title("Monthly sales and Temperature")
fig.tight_layout()




plt.show()
```

##### twiny()

```python

import matplotlib.pyplot as plt

days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5"]
distance = [50, 80, 60, 100, 120]
fuel = [4, 6, 5, 8, 10]

# Create figure and first axis
fig, ax1 = plt.subplots(figsize=(9, 5))

# Plot distance on the bottom X-axis
line1 = ax1.plot(
    distance, days,
    color="blue", marker="o", label="Distance (km)"
)

ax1.set_xlabel("Distance (km)", color="blue")
ax1.set_ylabel("Days")
ax1.tick_params(axis="x", labelcolor="blue")

# Create second X-axis at the top
ax2 = ax1.twiny()

# Plot fuel consumption on the top X-axis
line2 = ax2.plot(
    fuel, days,
    color="red", marker="s", label="Fuel (litres)"
)

ax2.set_xlabel("Fuel Consumption (litres)", color="red")
ax2.tick_params(axis="x", labelcolor="red")

# Combine legends
lines = line1 + line2
labels = [line.get_label() for line in lines]
ax1.legend(lines, labels, loc="lower right")

plt.title("Daily Distance and Fuel Consumption")
fig.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
hours=[1,2,3,4,5]
celsius=[0,10,20,30,40]
fig,ax=plt.subplots(figsize=(9,5))
ax.plot(hours,celsius,color="blue",marker="o")
ax.set_xlabel("Time(hours)")
ax.set_ylabel("Temperature(deg C)",color="blue")
ax.tick_params(axis="y",labelcolor="blue")
def celsius_to_far(c):
    return (c*9/5)+32

def far_to_celsius(f):
    return (f-32)*5/9

secax = ax.secondary_yaxis(
    "right",
    functions=(celsius_to_far, far_to_celsius)
)

secax.set_ylabel("Temperature (°F)", color="red")
secax.tick_params(axis="y", labelcolor="red")


plt.title("Temperature in celsius and fahrenheit")
fig.tight_layout()
plt.show()

```

#### logarithmic scale

#### symmetric logarithmic
* symlog- useful when dataset contains negative or zero values.

```python
import matplotlib.pyplot as plt
months=["Jan","Feb","Mar","Apr"]
sales=[100,150,120,200]
profit=[20,35,25,50]
fig,ax=plt.subplots(figsize=(10,4))
ax.plot(months,sales,marker="o",color="blue")
ax.plot(months,profit,marker="s",color="green")
labels=["Monthly sales","Monthly profit"]

ax.set_xlabel("Months")
ax.set_ylabel("Amount")
ax.set_title("Monthly Sales and Profit")

ax.legend(labels)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
fig,ax=plt.subplots()
ax.plot([1,2,3,4],[10,20,15,25],color="blue")
ax.axhspan(12,18,color="orange",alpha=0.3)
line_handle=Line2D([0],[0],color="blue",linewidth=2)
shade_handle=Patch(facecolor="orange",alpha=0.3)
ax.legend(handles=[line_handle,shade_handle],labels=["Observed Values","Target Range"])
ax.set_title("Chart with proxy legend entries")
ax.set_xlabel("X Values")
ax.set_ylabel("Y Values")
plt.show()
```

```python
import matplotlib.pyplot as plt  
fig,ax1=plt.subplots()

line1=ax1.plot([1,2,3,3],[10,20,15,25],color="blue",label="Tedmperature")

ax1.set_xlabel('Time')
ax1.set_ylabel('temperature (%)',color='blue')
ax2=ax1.twinx()
line2,=ax2.plot(
  [1,2,3,4],
  [50,60,55,70],
  color='red',
  linestyle='--',
  label='Humindity'  
)
ax2.set_ylabel('Humindity(%)',color='red')
handles1,labels1=ax1.get_legend_handles_labels()
handles2,labels2=ax2.get_legend_handles_labels()
ax1.legend(
    handles1+handles2,
    labels1+labels2,
    loc='upper left'
)
ax1.set_title('temperature and humindity')
plt.show()
```

```python
import matplotlib.pyplot as plt 
from matplotlib.ticker import MultipleLocator 

fig,ax=plt.subplots()
ax.plot([1,2,3,4,5],[10,20,30,40,50])
ax.xaxis.set_major_locator(MultipleLocator(2))
plt.show()
```

```python
import matplotlib.pyplot as plt 
from matplotlib.ticker import StrMethodFormatter

fig,ax=plt.subplots()
ax.plot([1,2,3,4],[1000,2500,3500,5000])
ax.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))
plt.show()
```
