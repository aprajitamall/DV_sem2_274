#### Vectorization
* vectorization means performing operations on entire arrays at once instead of using Python Loops.
* It is fast as numpy uses optimizes  C code internally.
* work on same shape array

```python
arr=np.arange(100000)
lst=[x for x in range(100000)]
# python list
start=time.time()
lst2=[x*2 for x in lst]
end =time.time()
print (end-start)
# numpy array
start1=time.time()
arr2=arr*2  # broadcasting
end1 =time.time()
print (end1-start1)

```

### Key Concepts
* Fancy Indexing vs. Slicing: Basic slicing (like arr[1:3]) returns a view where base points to the original array data. Fancy indexing (passing a list or array of indices) constructs a new, unlinked array.
* f.base Property: For a view, .base references the original memory buffer. For an independent copy created by fancy indexing, .base evaluates to None

### Advance  Ufunc  features
* Specifing output- for large calculations

```python
x=np.arange(2,20,2)
print(x)
z=np.empty(9)
print(z)
print(np.multiply(x,10,out=z)) # the resultant will be stored in z
```

 * Aggregations (Reduce method):if you like to reduce an array with a particular operations we can use the reduce method of any ufunc.
A reduce repeatedly applies a given operation to the elements of an array until only a single result remains.

```python
#alling reduce on the add ufunc returns the sum of all elements in the array
x=np.arange(1,11)
print(x)
np.add.reduce(x)
```

```python
np.add.accumulate(x) ## cummulative sum
```

```python
import numpy as np
x=np.array([1,2,4,np.nan])
y=np.array([9,7,6,5])
print(np.add.reduce(x))
print(np.sum(x))
print(np.prod(x))
print(np.nanprod(x)) ## it will ignore nan i.e not a number
print(np.nansum(x))
print(np.add.reduce(y))
print(np.sum(y))
print(np.prod(y))

```

```python
import numpy as np
x=np.arange(1,101).reshape(10,10)
print(x)
print(x[:,[1]])
```

```python
import numpy as np
x=np.arange(1,101).reshape(10,10)
print(x)
print(x[:,[1]])
print("sum of elements",np.sum(x[:,[1]]))
print("mean of elements",np.mean(x[:,[1]]))
print("Standard deviation",np.std(x[:,[1]]))
print(np.var(x[:,[1]]))
print("Product of elements",np.prod(x[:,[1]]))
print("Minimum value",np.min(x[:,[1]]))
print("Maximum value",np.max(x[:,[1]]))
print("Median",np.median(x[:,[1]]))
print("Any elements are true",np.any(x[:,[1]]))  # return true or false
print(np.all(x[:,[1]])) # evaluate all element are true

```
